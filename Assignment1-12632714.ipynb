```python
# 作业2 Matrix multiplication

# 2.1 生成两个随机矩阵M1和M2
import random
# M1是一个5行10列的矩阵
M1 = []
# 循环创建每一行，一共5行
for i in range(5):
    m1row = []
    # 一行里面使用random函数生成10个随机数
    for j in range(10):
        random_m1 = random.randint(0, 50)
        m1row.append(random_m1)
    M1.append(m1row)
# M2是一个10行5列的矩阵
M2 = []
# 循环创建每一行，一共10行
for i in range(10):
    m2row = []
    # 一行里面使用random函数生成5个随机数
    for j in range(5):
        random_m2 = random.randint(0, 50)
        m2row.append(random_m2)
    M2.append(m2row)

# 打印M1和M2看看结果
print("以下是矩阵M1(5行10列)")
for r in M1:
    print(r)
print("\n以下是矩阵M2(10行5列)")
for r in M2:
    print(r)

# 2.2 编写矩阵乘法函数Matrix_multip
def Matrix_multip(A, B):
    # 获取A的行数
    rows_A = len(A)
    # 获取B的列数
    cols_B = len(B[0])
    # 获取A的列数（等于B的行数）
    cols_A = len(A[0])
    # 创建结果矩阵，结果矩阵的大小应为rows_A×cols_B
    result = []
    # 外层循环：遍历A的每一行
    for i in range(rows_A):
        new_row = []
        # 中层循环：遍历B的每一列
        for j in range(cols_B):
            # 将乘机依次相加，初始的和值为0
            sum0 = 0
            # 内层循环：计算A第i行和B第j列的乘积
            for k in range(cols_A):
                sum0 = sum0 + A[i][k] * B[k][j]
            new_row.append(sum0)
        result.append(new_row)
    return result

# 调用函数计算M1*M2
M_result = Matrix_multip(M1, M2)

# 输出相乘后的矩阵
print("\n以下是相乘后的矩阵结果M1 * M2")
for row in M_result:
    print(row)
```

```python
# 作业3 Pascal triangle
# 据查，帕斯卡三角是从第0行开始的，第0行是[1]，第1行[1,1]，那第k行对应的列表索引正好为k

def Pascal_triangle(k):
    # 第0行即初始行只有数字1
    record_row = [1]
    # 外循环：从第1行一直构建到第k行，每次构建都利用上一行的数据算出新一行，range左闭右开
    for i in range(1, k + 1):
        # 每一行第一个数字永远是1
        new_row = [1]
        # 遍历上一行，把相邻两个数字相加，得到中间的值，正好上一行长度是i，所以中间元素循环到i-1
        # 当i=1时，对应帕斯卡三角形第1行[1,1]，中间没有元素，不需要相加，正好符合逻辑
        for j in range(1, i):
            sum0 = record_row[j - 1] + record_row[j]
            new_row.append(sum0)
        # 每一行最后一个数字永远是1
        new_row.append(1)
        # 更新record_row，把构建好的new_row赋值给record_row
        record_row = new_row
    # 循环结束，record_row就是第k行，打印这一行
    print(f"帕斯卡三角形第{k}行的结果如下:")
    print(record_row)

# 输出帕斯卡三角形第100行和第200行的数据
print("以下是帕斯卡三角形第100行的数据：")
Pascal_triangle(100)

print("\n以下是帕斯卡三角形第200行的数据：")
Pascal_triangle(200)
```

```python
#助教添加
def Pascal_Triangle(k):
    Row = [1]

    for n in range(1, k):
        New_Row = [1] * (n + 1)

        for i in range(1, n):
            New_Row[i] = Row[i - 1] + Row[i]

        Row = New_Row

    return Row


print(Pascal_Triangle(100))
print(Pascal_Triangle(200))
```

```python
# 作业4 Add or double

def Least_moves(x):
    # 将步数初始化为0
    move = 0
    # 一开始将目标金额x赋值给当前金额开始往回倒推到1，算倒推步数
    current = x
    # 只要current还大于1就继续倒推
    while current > 1:
        # 判断当前数字是不是奇数
        if current % 2 == 1:
            # 奇数：最后一步只能是add，所以倒推就减1
            current = current - 1
        else:
            # 偶数：最后一步只能是double，倒推就除以2
            current = current // 2
        # 每做一次倒推操作，步数+1
        move = move + 1
    # 当current == 1时循环结束 ，move就是最小步数
    print(move)

# 测试作业里给出的给的例子
print("Test_Least_moves(2):")
Least_moves(2)
print("Test Least_moves(5):")
Least_moves(5)

# 随机选1~100之间的整数测试
import random
num = random.randint(1, 100)
print(f"取到的随机数为{num},它的对应最小步数应该是应该是:")
Least_moves(num)
```

```python
# 作业5 Dynamic programming
# 参考博客的网址如下
# https://blog.csdn.net/qq_46450354/article/details/127183649?ops_request_misc=elastic_search_misc&request_id=74d0d2278d4ee937581fbc77dbd2be28&biz_id=0&utm_medium=distribute.pc_search_result.none-task-blog-2~all~top_positive~default-2-127183649-null-null.142^v102^pc_search_result_base8&utm_term=python%20eval%EF%BC%88%EF%BC%89&spm=1018.2226.3001.4187
# https://xzknet.blog.csdn.net/article/details/102504635?spm=1001.2101.3001.6650.4&utm_medium=distribute.pc_relevant.none-task-blog-2%7Edefault%7EBlogCommendFromBaidu%7ERate-4-102504635-blog-45672197.235%5Ev43%5Epc_blog_bottom_relevance_base4&depth_1-utm_source=distribute.pc_relevant.none-task-blog-2%7Edefault%7EBlogCommendFromBaidu%7ERate-4-102504635-blog-45672197.235%5Ev43%5Epc_blog_bottom_relevance_base4
# 绘图部分有询问借助AI
# 5.1 打印1~100任意整数的解的算式
def Find_expression(target):
    digits = "123456789"
    # 用来保存找到的该整数的解的算式
    result_list = []

    # pos:当前处理到第几个数字，current:当前拼接出来的表达式字符串如"1-2+34"
    def dfs(pos, current):
        # 当已经处理完所有9个数字，则计算完一种情况，判断解是否正确后跳出循环
        if pos == len(digits):
            # 计算表达式的值（函数内部自我递归调用，只能pycharm强制要求try-catch语句）
            try:
                val = eval(current)
            except:
                return
            # 判断是否等于目标值，等于的话则加入结果算式
            if val == target:
                result_list.append(current)
            return

        # 从pos位置开始，尝试向后拼接数字
        # 最多可以一直拼到最后一位
        for i in range(pos, len(digits)):
            # 取出一段数字字符串
            num_str = digits[pos:i+1]
            # 第一种分支：在当前数字前面加 '+'
            dfs(i+1, current + "+" + num_str)
            # 第二种分支：在当前数字前面加 '-'
            dfs(i+1, current + "-" + num_str)

    # 循环递归起点：第一个数字固定是digits[0] = 1，从pos=1开始往后填符号
    #dfs(1, digits[0])
    for first_end in range(len(digits)):#助教修改：原代码固定从数字1开始并立即添加正负号，遗漏了以12等连续数字开头的表达式
        dfs(first_end + 1, digits[:first_end + 1])#助教修改
    return result_list

# 测试调用5.1函数，找所有等于50、60、70的表达式
print("\n测试50的所有表达式，输出Find_expression(50)")
Find_expression(50)
ans = Find_expression(50)
for exp in ans:
    print(f"{exp} = 50")

print("\n测试60的所有表达式，输出Find_expression(60)")
Find_expression(60)
ans = Find_expression(60)
for exp in ans:
    print(f"{exp} = 60")

print("\n测试70的所有表达式，输出Find_expression(70)")
Find_expression(70)
ans = Find_expression(70)
for exp in ans:
    print(f"{exp} = 70")


# 5.2 统计1~100每个数字的解数量，绘图
import matplotlib.pyplot as plt

# 初始化列表，Total_solutions[i]存放值等于(i+1)的解数量
Total_solutions = []

# 遍历目标 i 从1到100
for target_num in range(1, 101):
    # 调用函数获取所有解
    solutions = Find_expression(target_num)
    # 把解的个数存入列表
    Total_solutions.append(len(solutions))

# 绘图部分
plt.figure(figsize=(12,5))
# x轴：1~100；y轴：对应解的数量
plt.plot(range(1,101), Total_solutions, marker='o', markersize=2)
plt.xlabel("Target value (1~100)")
plt.ylabel("Number of solutions")
plt.title("Count of expressions from 123456789 for each target 1~100")
plt.grid(True, alpha=0.3)
plt.show()


# 找出最大值、最小值对应的数字
# 最大解的数量
max_count = max(Total_solutions)
# 所有拥有最大数量解的目标数字
max_nums = [i+1 for i, cnt in enumerate(Total_solutions) if cnt == max_count]

# 最小解的数量
min_count = min(Total_solutions)
# 所有拥有最小数量解的目标数字
min_nums = [i+1 for i, cnt in enumerate(Total_solutions) if cnt == min_count]

print("\n结果统计")
print(f"最多的解的数量 = {max_count}, 对应的数字为: {max_nums}")
print(f"最少的解的数量 = {min_count}, 对应的数字为: {min_nums}")
```
